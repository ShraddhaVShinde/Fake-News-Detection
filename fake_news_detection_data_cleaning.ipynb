```python
import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
# --- STEP 1: Load Dataset ---
df = pd.read_csv("Downloads/WELFake_dataset/WELFake_Dataset.csv")
df.head()
```

```python
# --- STEP 2: Handle Missing Values & Duplicates ---
# Remove any rows with null values to avoid errors during text processing 
df = df.dropna() 

df = df.drop_duplicates()
```

```python
# --- STEP 3: Combine Columns ---
# Combine 'title' and 'text' into a single 'content' column 
df['content'] = df['title'] + " " + df['text']
df.head()
 
df = df[['content', 'label']]
df.head()
```

```python
# --- STEP 4: NLP Preprocessing ---
nltk.download('stopwords')
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Lowercasing: Convert text to lowercase
    text = text.lower()
    
    # 2. Remove Punctuation & Numbers: Keep only alphabets
    text = re.sub('[^a-zA-Z]', ' ', text)
    
    # 3. Tokenization: Split text into individual words
    words = text.split()
    
    # 4. Remove Stopwords & Stemming: Filter common words and get root forms
    # Example: "running" -> "run"
    cleaned_words = [ps.stem(word) for word in words if word not in stop_words]
    
    # Join the words back into a single string
    return " ".join(cleaned_words)
```

```python
# --- STEP 5: Apply Cleaning ---

df['content'] = df['content'].apply(preprocess_text)
```

```python
# --- STEP 6: Save Cleaned Data ---
df.to_csv("cleaned_news_welfake_data.csv", index=False)

print("Data Cleaning Complete. File saved as 'cleaned_news_welfake_data.csv'.")
```
